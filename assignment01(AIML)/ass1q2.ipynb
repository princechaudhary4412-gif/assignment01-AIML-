**Consider a binomial distribution. Assume that a product is either rated positive or negative in an ecommerce website. You got 97 positive reviews out of 100 for a product. Construct a binomial distribution with different success rates to find out an estimated success rate for the product. Use histograms to assert your answer.**

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Given data
n = 100
observed_positive = 97

# Different success rates to test
success_rates = [0.90, 0.93, 0.95, 0.97, 0.99]

# Number of simulations
experiments = 10000

# Set random seed for reproducible results
np.random.seed(42)

# Create figure
plt.figure(figsize=(15, 10))

# Simulate for each success rate
for i, p in enumerate(success_rates):

    # Generate binomial results
    results = np.random.binomial(n, p, experiments)

    # Create subplot
    plt.subplot(2, 3, i + 1)

    # Create histogram
    plt.hist(results, bins=np.arange(50, 101) - 0.5, edgecolor='black')

    # Mark the observed value
    plt.axvline(observed_positive, linestyle='--', linewidth=2, label='Observed = 97')

    # Add title and labels
    plt.title(f"Success Rate p = {p}")
    plt.xlabel("Number of Positive Reviews")
    plt.ylabel("Frequency")
    plt.legend()

# Adjust the spacing
plt.tight_layout()

# Display the histograms
plt.show()
```
