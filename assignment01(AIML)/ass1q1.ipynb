**Let us consider a random experiment of rolling a die variable number of times (each die roll can be denoted as \(X_1, X_2, \ldots, X_{50}\) ranging from 1 to 50) and collecting their sum (No need to collect the sum when you roll the die once).

Repeat the experiments 100, 500, 1000, 5000, 10000, 50000, and 100000 times.

Plot and save the histograms of each.
(For example: you should have 6 histograms for one die roll, 6 for 2 dice roll, 6 for 3 dice roll, and so on).

Find the mean and variance for each case.**

```python
from IPython.display import Image, display

display(Image(filename="/content/Screenshot_20260903_112433.png"))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# Number of experiments
experiments = [100, 500, 1000, 5000, 10000, 50000, 100000]

# Create folder to save histograms
os.makedirs("histograms", exist_ok=True)

# Store all results
results = []

# Roll dice from 1 to 50
for n in range(1, 51):

    # Repeat the experiment for different values of N
    for N in experiments:

        # Generate random dice rolls
        # N = number of experiments
        # n = number of dice in each experiment
        rolls = np.random.randint(1, 7, size=(N, n))

        # Find sum of dice for every experiment
        sums = np.sum(rolls, axis=1)

        # Calculate experimental mean
        mean = np.mean(sums)

        # Calculate experimental variance
        variance = np.var(sums)

        # Calculate theoretical mean
        theoretical_mean = 3.5 * n

        # Calculate theoretical variance
        theoretical_variance = (35 / 12) * n

        # Store the results
        results.append([
            n,
            N,
            mean,
            variance,
            theoretical_mean,
            theoretical_variance
        ])

        # Create histogram
        plt.figure(figsize=(8, 5))

        plt.hist(
            sums,
            bins="auto",
            edgecolor="black"
        )

        # Title and labels
        plt.title(
            f"Sum of {n} Dice Rolls - {N} Experiments"
        )

        plt.xlabel("Sum of Dice")
        plt.ylabel("Frequency")

        plt.grid(
            axis="y",
            alpha=0.3
        )

        # Name of histogram file
        filename = f"histograms/dice_{n}_experiments_{N}.png"

        # Save histogram
        plt.savefig(
            filename,
            dpi=150
        )

        # Close the figure
        plt.close()


# Print message after all histograms are created
print("\nAll histograms have been generated and saved.")


# Print results heading
print("\n")
print(
    "n\tN\tMean\t\tVariance\tTheoretical Mean\tTheoretical Variance"
)


# Print every result
for row in results:

    print(
        f"{row[0]}\t"
        f"{row[1]}\t"
        f"{row[2]:.4f}\t\t"
        f"{row[3]:.4f}\t\t"
        f"{row[4]:.4f}\t\t"
        f"{row[5]:.4f}"
    )
```
