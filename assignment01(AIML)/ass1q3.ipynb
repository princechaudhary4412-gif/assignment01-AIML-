**For the dataset given in advertising.csv, find out the entropy for each column. [Hint: If it is a categorical variable, then it is straightforward. If it is a continuous value, try binning it and then find entropy.]**

```python
##calculating entropy


import pandas as pd
import numpy as np

# Load your advertising.csv file
df = pd.read_csv("advertising.csv")

# Function to calculate entropy
def entropy(column):
    probabilities = column.value_counts(normalize=True)
    return -np.sum(probabilities * np.log2(probabilities))

# Calculate entropy for every column
for col in df.columns:

    # Divide continuous values into 5 bins
    binned = pd.qcut(df[col], q=5, duplicates='drop')

    # Calculate entropy
    ent = entropy(binned)

    print(f"Entropy of {col}: {ent:.4f}")
```
